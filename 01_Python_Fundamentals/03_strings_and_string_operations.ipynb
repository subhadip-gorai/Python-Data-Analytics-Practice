```python
# Extracting multiple elements
```

```python
b = 'India is my country'

# Extract 5 characters

b[0:5]  # 0 is a starting index and 5 is a ending index

# Ending index is excluded
```

```python
b[4:9]
```

```python
b = 'India is my country.'

# Extract last 4 characters

b[-4:-1]  # last 4 characters are not showing because Ending index is excluded
```

```python
b[-4:]  # last 4 characters are showing because after colon is empty.
# This empty index position define end of the string
```

```python
b = 'India is my country.'

# Extract first 5 characters

b[:5]  # This empty index position define start of the string
```

```python
b[2:]
```

```python
b[0:5:1]  # After 2nd colon this number define step size of elements.
```

```python
b[0:5:]  # Its by default 1 and this 1 means no skip elements.
```

```python
b[0:5:2]  # Skip elements using step size and step size 2 means 1 element gap.
```

```python
b[0:10:3]  # step size 3 means 2 element gap.

# It means step size is n than element gap (n-1)
```

```python
b[:]  # Empty index define start from the starting index to ending index
```

```python
b[::2]  # All string with skip elements using step size.
```

```python
c = '123456789'

# Extract odd characters
c[::2]
```

```python
# Extract even characters
c[1::2]
```
